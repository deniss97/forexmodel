```python
### Функция для микроцикла

def download_moex_candles(
    ticker: str,
    start_date: str = None,
    end_date: str = None,
    interval: str = '1min',
    filename: str = None,
    max_attempts: int = 3,
    min_rows_per_request: int = 10000   # ← новый параметр
) -> str:
    """
    Скачивает свечи по тикеру с защитой от таймаутов.
    
    Новая логика:
    - Если в итерации загружено меньше min_rows_per_request (по умолчанию 10000),
      то цикл завершается (считаем, что дальше данных в этом диапазоне нет).
    - Новая итерация начинается с максимальной даты включительно.
    """
    if end_date is None:
        end_date = datetime.today().strftime('%Y-%m-%d')
    
    if filename is None:
        filename = f"{ticker}_{interval}.csv"

    temp_filename = filename.replace('.csv', '.temp.csv')

    t = Ticker(ticker)
    target_end = pd.to_datetime(end_date).date()

    # Определяем стартовую дату
    if os.path.exists(filename) and os.path.getsize(filename) > 0:
        df_exist = pd.read_csv(filename)
        last_date = pd.to_datetime(df_exist['begin']).max().date()
        next_start = last_date
        print(f"✅ Продолжаем выгрузку с {next_start} (включительно)")
    else:
        next_start = pd.to_datetime(start_date).date() if start_date else None
        print(f"📁 Создаётся новый файл: {filename}")

    total_downloaded = 0
    iteration = 0

    while True:
        iteration += 1
        start_str = next_start.strftime('%Y-%m-%d') if next_start else None
        
        print(f"\n🔄 Итерация {iteration}: {start_str or 'начало истории'} → {end_date} | {interval}")

        success = False
        for attempt in range(1, max_attempts + 1):
            try:
                df = t.candles(
                    period=interval,
                    start=start_str,
                    end=end_date
                )
                success = True
                break
            except Exception as e:
                if "ConnectTimeout" in str(e) or "timeout" in str(e).lower():
                    pause = 5 + attempt * 8
                    print(f"   ⚠️ ConnectTimeout (попытка {attempt}/{max_attempts}). Ждём {pause} сек...")
                    time.sleep(pause)
                else:
                    print(f"   ❌ Неизвестная ошибка: {type(e).__name__}: {e}")
                    raise

        if not success:
            print("❌ Не удалось получить данные после нескольких попыток.")
            break

        if df is None or len(df) == 0:
            print("✅ Новых данных нет.")
            break

        # Сохраняем данные во временный файл
        if not os.path.exists(temp_filename):
            df.to_csv(temp_filename, index=False)
        else:
            df.to_csv(temp_filename, mode='a', header=False, index=False)

        total_downloaded += len(df)

        current_max = pd.to_datetime(df['begin']).max().date()
        rows_loaded = len(df)

        print(f"   → Загружено {rows_loaded:,} свечей | Макс. дата: {current_max}")

        # === НОВАЯ ЛОГИКА: если строк меньше 10000 — завершаем цикл ===
        if rows_loaded < min_rows_per_request:
            print(f"   ⚠️ Загружено менее {min_rows_per_request:,} строк ({rows_loaded:,}). "
                  f"Считаем, что данные закончились. Завершаем выгрузку.")
            break

        # Проверяем, достигли ли целевой даты
        if current_max >= target_end:
            print("🎉 Достигнута целевая дата окончания!")
            break

        # Обновляем дату для следующей итерации
        next_start = current_max

        # Пауза перед следующей итерацией
        pause = random.randint(5, 20)
        print(f"⏳ Пауза {pause} секунд перед следующей итерацией...")
        time.sleep(pause)

    # === Финальная очистка от дублей ===
    print("\n🧹 Очистка итогового файла от дубликатов...")
    if os.path.exists(temp_filename):
        df_final = pd.read_csv(temp_filename)
        df_final['begin'] = pd.to_datetime(df_final['begin'])
        df_final = df_final.drop_duplicates(subset=['begin'], keep='last')
        df_final = df_final.sort_values(by='begin').reset_index(drop=True)
        
        df_final.to_csv(filename, index=False)
        os.remove(temp_filename)

        print(f"   Строк после очистки: {len(df_final):,}")
    else:
        print("⚠️ Нет данных для сохранения.")

    print(f"\n✅ Выгрузка завершена!")
    print(f"   Тикер: {ticker} | Интервал: {interval}")
    print(f"   Период: {start_date or 'начало истории'} — {end_date}")
    print(f"   Всего загружено: {total_downloaded:,} свечей")
    print(f"   Итоговый файл: {filename}")

    return filename
```

```python

def download_moex_full_period(
    ticker: str,
    start_date: str,
    end_date: str,
    interval: str = '1min',
    output_folder: str = "data",
    months_per_chunk: int = 1,
    min_rows_per_request: int = 10000
):
    """
    Улучшенный макроцикл для скачивания данных по подпериодам.
    Теперь правильно продолжает выгрузку с последней скачанной даты внутри каждого чанка.
    """
    os.makedirs(output_folder, exist_ok=True)

    # Генерируем начало каждого чанка (по месяцам)
    chunk_starts = pd.date_range(start=start_date, end=end_date, freq=f'{months_per_chunk}MS')

    print(f"🚀 Запуск улучшенного макроцикла для {ticker} | {interval}")
    print(f"   Период: {start_date} — {end_date}")
    print(f"   Разбиение: по {months_per_chunk} месяц{'у' if months_per_chunk == 1 else 'ев'}\n")

    for i in range(len(chunk_starts)):
        chunk_start_dt = chunk_starts[i]
        sub_start = chunk_start_dt.strftime('%Y-%m-%d')

        # Определяем конец текущего чанка
        if i + 1 < len(chunk_starts):
            chunk_end_dt = chunk_starts[i + 1] - timedelta(days=1)
        else:
            chunk_end_dt = pd.to_datetime(end_date)

        sub_end = min(chunk_end_dt, pd.to_datetime(end_date)).strftime('%Y-%m-%d')

        filename = f"{ticker}_{interval}_{sub_start}_{sub_end}.csv"
        full_path = os.path.join(output_folder, filename)

        print(f"\n📌 Чанк {i+1}/{len(chunk_starts)}: {sub_start} — {sub_end}")

        # === Улучшенная проверка существующего файла ===
        resume_start = None
        if os.path.exists(full_path) and os.path.getsize(full_path) > 10000:  # файл выглядит осмысленным
            try:
                df_exist = pd.read_csv(full_path)
                if 'begin' in df_exist.columns:
                    last_date = pd.to_datetime(df_exist['begin']).max().date()
                    target_end = pd.to_datetime(sub_end).date()

                    if last_date >= target_end:
                        print(f"✅ Чанк полностью скачан до {last_date}")
                        continue
                    else:
                        resume_start = (last_date + timedelta(days=1)).strftime('%Y-%m-%d')
                        print(f"   Продолжаем с {resume_start} (уже есть до {last_date})")
                else:
                    print("   ⚠️ Файл существует, но структура некорректна — перезаписываем")
            except Exception as e:
                print(f"   ⚠️ Не удалось прочитать файл ({e}), будет перезаписан")

        try:
            download_moex_candles(
                ticker=ticker,
                start_date=resume_start or sub_start,   # ← ключевой момент
                end_date=sub_end,
                interval=interval,
                filename=full_path,
                min_rows_per_request=min_rows_per_request
            )
        except Exception as e:
            print(f"❌ Ошибка в чанке {sub_start}—{sub_end}: {e}")
            print("   Макроцикл остановлен. При следующем запуске продолжит с этого места.")
            break

        # Пауза между чанками
        pause = random.randint(10, 35)
        print(f"⏳ Пауза {pause} секунд перед следующим чанком...")
        time.sleep(pause)

    print(f"\n🎉 МАКРО-ВЫГРУЗКА ЗАВЕРШЕНА!")
    print(f"   Все файлы сохранены в папке: {output_folder}")
    print(f"   Тикер: {ticker} | Интервал: {interval}")
```

# Скачивание фьючей

```python

def download_moex_futures_full_period(
    ticker_prefix: str,           # 'BR', 'GD', 'SI', 'GZ' и т.д.
    start_date: str,
    end_date: str = None,
    interval: str = '1min',
    output_folder: str = "futures_data",
    months_per_chunk: int = 1
):
    """
    Скачивает все фьючерсы по заданному префиксу (например все BR*)
    и сохраняет каждый контракт в отдельный файл.
    """
    if end_date is None:
        end_date = datetime.now().strftime('%Y-%m-%d')
    
    os.makedirs(output_folder, exist_ok=True)
    
    print(f"🚀 Запуск выгрузки фьючерсов по префиксу '{ticker_prefix}'")
    print(f"Период: {start_date} — {end_date} | Интервал: {interval}\n")
    
    # Генерируем начало чанков
    chunk_starts = pd.date_range(start=start_date, end=end_date, freq=f'{months_per_chunk}MS')
    
    for i in range(len(chunk_starts)):
        chunk_start_dt = chunk_starts[i]
        sub_start = chunk_start_dt.strftime('%Y-%m-%d')
        
        if i + 1 < len(chunk_starts):
            chunk_end_dt = chunk_starts[i + 1] - timedelta(days=1)
        else:
            chunk_end_dt = pd.to_datetime(end_date)
        
        sub_end = min(chunk_end_dt, pd.to_datetime(end_date)).strftime('%Y-%m-%d')
        
        print(f"\n📌 Чанк {i+1}/{len(chunk_starts)}: {sub_start} — {sub_end}")
        
        # Получаем список всех активных фьючерсов с нужным префиксом
        try:
            all_futures = Market("futures").tickers()
            active_futures = all_futures[all_futures['ticker'].str.startswith(ticker_prefix, na=False)]
            
            if active_futures.empty:
                print(f"   Нет фьючерсов с префиксом {ticker_prefix}")
                continue
                
            print(f"   Найдено контрактов: {len(active_futures)}")
            
            for _, row in active_futures.iterrows():
                ticker = row['ticker']
                filename = f"{ticker}_{interval}_{sub_start}_{sub_end}.csv"
                full_path = os.path.join(output_folder, filename)
                
                # Проверка существующего файла
                if os.path.exists(full_path) and os.path.getsize(full_path) > 50000:
                    print(f"   ✅ {ticker} уже скачан")
                    continue
                
                print(f"   ↓ Скачиваем {ticker}")
                
                try:
                    instr = Ticker(ticker)
                    data = instr.candles(
                        start=sub_start,
                        end=sub_end
                    )
                    
                    if not data.empty:
                        data = data.reset_index()
                        data.to_csv(full_path, index=False)
                        print(f"     ✓ Сохранено {len(data)} строк")
                    else:
                        print(f"     ⚠️ Нет данных для {ticker}")
                        
                except Exception as e:
                    print(f"     ❌ Ошибка {ticker}: {e}")
                
                time.sleep(random.uniform(1.5, 3.5))  # анти-бан
                
        except Exception as e:
            print(f"❌ Ошибка при получении списка фьючерсов: {e}")
            continue
    
    print(f"\n🎉 Выгрузка фьючерсов по префиксу '{ticker_prefix}' завершена!")
    print(f"Данные сохранены в папку: {output_folder}")
```
